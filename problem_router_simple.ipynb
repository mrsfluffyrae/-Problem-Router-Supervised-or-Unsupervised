## The rule I use
For every problem I ask three questions:

1. **Label:** do we have known correct answers from the past?
2. **Duty:** do we want to *predict* that answer for new cases, or *discover* patterns?
3. **Output:** what do we get back (a category, a number, a group, a score, or rules)?

**Supervised** = a label exists AND we want to predict it. **Unsupervised** = otherwise.

## Bonus: an ambiguous case (Case 10, support tickets)

**For supervised:** the company already has categories and 8,000 tagged tickets, and a new ticket must be sent to a named team.

**For unsupervised:** only 4% of tickets are tagged, the tags may be inconsistent, and clustering could find new issue types.

## Task 5: Demonstrations

### Supervised: there IS a `y` (known answers) and we can score it

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

rng = np.random.default_rng(42)
attendance = rng.uniform(40, 100, 300)
quiz = rng.uniform(20, 100, 300)
X = pd.DataFrame({"attendance": attendance, "quiz": quiz})
y = ((0.5*attendance + 0.5*quiz + rng.normal(0, 8, 300)) >= 60).astype(int)   # known pass/fail

X_tr, X_te, y_tr, y_te = train_test_split(X, y, random_state=0)
model = LogisticRegression().fit(X_tr, y_tr)
print("Accuracy on unseen students:", round(model.score(X_te, y_te), 2))
print("New student (55% attendance, 48 quiz) passes?", model.predict(pd.DataFrame({"attendance":[55],"quiz":[48]}))[0] == 1)
```

### Unsupervised: there is NO `y`, we only get groups to interpret

```python
from sklearn.cluster import KMeans

rng = np.random.default_rng(7)
logs = pd.DataFrame({
    "hours_online": np.r_[rng.normal(2,.6,100), rng.normal(9,1.2,100), rng.normal(6,1,100)],
    "forum_posts":  np.r_[rng.normal(1,.8,100), rng.normal(2,1,100),  rng.normal(14,2.5,100)],
})
logs["group"] = KMeans(n_clusters=3, n_init=10, random_state=0).fit_predict(logs[["hours_online","forum_posts"]])
print(logs.groupby("group").mean().round(1))
print("\nNo accuracy here: a person must read the averages and name each group.")
```

**Peer swap test:** a swapped case still works with the same three questions.
- "Predict which customers will cancel" -> past cancellations exist and we predict them -> **Supervised**.
- "Group songs by sound, no genre tags" -> no label, we discover groups -> **Unsupervised**.
